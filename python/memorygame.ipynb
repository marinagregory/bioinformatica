```python
import random # sirve para generar numeros al azar al tirar el dado y elegir las preguntas
import sys # permite interactuar con el sistema, por ejemplo para cerrar el programa con sys.exit()


TAM_TABLERO, CARAS_DADO = 20, 6 # se escribe en mayúscculas para valores que no cambian

# Preguntas por tema en español, con opciones y respuesta correcta
TEMAS = {"Ciencia": [  {"preg": "¿Qué planeta es conocido como el Planeta Rojo?", "opts": ["Tierra", "Marte", "Júpiter", "Venus"], "resp": "Marte"},
        {"preg": "¿Qué gas absorben las plantas y exhalamos los humanos?", "opts": ["Oxígeno", "Nitrógeno", "Dióxido de carbono", "Hidrógeno"], "resp": "Dióxido de carbono"},
        {"preg": "¿Cuál es el símbolo químico del agua?", "opts": ["WO", "H2O", "HO2", "O2H"], "resp": "H2O"},
        {"preg": "¿Cuántos huesos tiene un adulto?", "opts": ["206", "195", "220", "180"], "resp": "206"},
        {"preg": "¿Qué fuerza nos mantiene en el suelo?", "opts": ["Magnetismo", "Gravedad", "Fricción", "Inercia"], "resp": "Gravedad"} ],
    "Historia": [  {"preg": "¿Quién fue el primer presidente de EE.UU.?", "opts": ["Abraham Lincoln", "George Washington", "Thomas Jefferson", "John Adams"], "resp": "George Washington"},
        {"preg": "¿En qué año terminó la Segunda Guerra Mundial?", "opts": ["1945", "1942", "1939", "1950"], "resp": "1945"},
        {"preg": "La Gran Pirámide está en qué país?", "opts": ["México", "Egipto", "India", "China"], "resp": "Egipto"},
        {"preg": "¿Quién escribió la Declaración de Independencia?", "opts": ["Benjamin Franklin", "George Washington", "Thomas Jefferson", "James Madison"], "resp": "Thomas Jefferson"},
        {"preg": "¿Qué muro dividió Berlín hasta 1989?", "opts": ["Gran Muralla China", "Muro de Berlín", "Muro de Adriano", "Muro de los Lamentos"], "resp": "Muro de Berlín"} ],
    "Deportes": [   {"preg": "¿Cuántos jugadores en un equipo de fútbol en el campo?", "opts": ["9", "10", "11", "12"], "resp": "11"},
        {"preg": "¿Qué deporte usa un volante llamado 'shuttlecock'?", "opts": ["Tenis", "Bádminton", "Golf", "Squash"], "resp": "Bádminton"},
        {"preg": "¿Qué país ganó la primera Copa Mundial de la FIFA?", "opts": ["Brasil", "Italia", "Uruguay", "Alemania"], "resp": "Uruguay"},
        {"preg": "¿Cuántos puntos vale un tiro libre en baloncesto?", "opts": ["1", "2", "3", "4"], "resp": "1"},
        {"preg": "¿Cómo se llama la puntuación cero en tenis?", "opts": ["Love", "Duck", "Nil", "Zero"], "resp": "Love"} ],
    "Geografía": [ {"preg": "¿Cuál es el río más largo del mundo?", "opts": ["Amazonas", "Nilo", "Yangtsé", "Misisipi"], "resp": "Nilo"},
        {"preg": "¿Qué país tiene la mayor población?", "opts": ["India", "Estados Unidos", "China", "Rusia"], "resp": "China"},
        {"preg": "¿En qué continente está Egipto?", "opts": ["Asia", "África", "Europa", "Sudamérica"], "resp": "África"},
        {"preg": "¿En qué cordillera está el Monte Everest?", "opts": ["Andes", "Alpes", "Himalayas", "Montañas Rocosas"], "resp": "Himalayas"},
        {"preg": "¿Cuál es el océano más pequeño?", "opts": ["Índico", "Pacífico", "Ártico", "Atlántico"], "resp": "Ártico"} ] }

def obtener_tema(pos): # def sirve para crear una función: un bloque de código con un nombre que puedes reutilizar cuando quieras
    if 1 <= pos <= 5: # if sirve para tomar decisiones: ejecuta un bloque de código solo si se cumple una condición
        return "Ciencia" 
    elif 6 <= pos <=10: # comprueba otra condición si la anterior no se cumplió
        return "Historia"
    elif 11 <= pos <=15:
        return "Deportes" # return sirve oara que la función entregue un resultado y termine. Cuando python llega a un return, la función se detiene y devuelve ese valor a quien la llamó
    else: # else (en caso contrario): se ejecuta cuando no se cumplió ninguna de las anteriores
        return "Geografía"

def bienvenida(): #define la función de bienvenida. Los paréntesis vacíos indican que no necesita recibir ningún dato
    print("\n===== Bienvenido al Juego de Memoria =====") # print muestra texto en pantalla y /n es un salto de línea que deja una línea en blanco antes del título para que se vea ordenador
    print("Tu objetivo es llegar a la casilla 20 lanzando un dado de 6 caras.")
    print("Cada vez que caes en una casilla, responderás una pregunta del tema correspondiente.")
    print("¡Si respondes correcto, avanzas! Si no, perderás ese turno.\n") # el /n del final deja otra línea en blanco para separa la bienvenida del resto del juego

def tirar_dado(): #define la función de tirar el dado
    return random.randint(1, CARAS_DADO) # elige un número al azar entre el 1 y 6. Return y dime cual ha salido

def preguntar(tema): #se define una función que reccibe un parámetro: el tema
    pregunta = random.choice(TEMAS[tema]) # accede al diccionario TEMAS, toma la lista de preguntas del tema y elige una al azar. El resultado se guarda en la variable pregunta, que es otro diccionario con tres claves: preg, opts y resp
    print(f"\nTema: {tema}\n{pregunta['preg']}") # print con f-string para insertar el tema y el texto de la pregunta. /n para el salto de línea
    for idx, opcion in enumerate(pregunta['opts'], 1): # para recorrer las opciones y numerarlas desde 1
        print(f"{idx}. {opcion}")
    while True: # bucle infinito que solo termina con un return. Garantiza que el jugador no avance hasta dar una respuesta válida
        respuesta = input("Selecciona el número de tu respuesta: ").strip() #lee la respuesta y quita espacios sobrantes
        if respuesta.isdigit() and 1 <= int(respuesta) <= len(pregunta['opts']): # 
            respuesta_int = int(respuesta) # int convierte el texto en número
            if pregunta['opts'][respuesta_int -1] == pregunta['resp']: # -1 porque los índices de las lsitas empiezan en 0, pero el jugador se le muestran desde 1
                print("¡Correcto!\n")
                return True # si es correcta
            else:
                print(f"Incorrecto. La respuesta correcta es: {pregunta['resp']}\n")
                return False # si es incorrecta
        else: # muestra un mensaje de entrada inválida y al no haber return, el while repite la pregunta
            print(f"Respuesta inválida. Ingresa un número entre 1 y {len(pregunta['opts'])}.")

def mostrar_tablero(pos): # función con un parámetro, pos, que es la posicción actual del jugador
    linea = "" # se crea una cadena de texto vacía donde se irá construyendo el dibujo de tablero
    for i in range(1, TAM_TABLERO +1): # recorre los números del 1 al 20. +1 porque range excluye el ultimo valor (1,21) va del 1 al 20. Usa la constante tam_tablero asi que si cambias el tamaño del tablero la función se adapta sola
        if i == pos: # si la casilla que se está dibujando es la el jugados, se añade X
            linea += "[X]"
        else:
            linea += f"[{i}]" # en cualquier otra casilla se añade su número entre corches con un f-string 1, 2, 3,...
    print("\nTablero:") # muestra el título con un salto de línea prevvio
    print(linea + "\n") # imprime todo el tablero en una sola línea, más un salto de línea final

def jugar():
    posicion = 1 # variable que guarda la casilla actual del jugador. Empieza en 1
    mostrar_tablero(posicion) # dibuja el tablero inicial
    while posicion < TAM_TABLERO: # el juego se repite mientras el jugador no haya llegado a la casilla 20. Cada vuelta del bucle es un turno
        input("Presiona Enter para lanzar el dado...") # pausa el programa hasta que el jugador pulse enter. Aquí  input se usa solo para esperar, no para guardar un dato
        dado = tirar_dado() # llama a la función que devuelve un número del 1 al 6
        print(f"Has sacado un {dado}.") # informa del resultado del dado
        nueva_pos = posicion + dado # se calcula a dónde iría el jugador, pero todavia no se mueve
        if nueva_pos > TAM_TABLERO: # si el dado lo llevaría más allá de la casilla 20, se muestra un aviso
            print("No puedes avanzar más allá de la casilla 20. Espera al siguiente turno.")
            continue # continue salta al inicio del bucle y comienza un nuevo turno, sin ejecutar el resto del código. El jugador se queda donde estaba. Esto significa que hay que sacar el número exacto para llegar a 20
        posicion = nueva_pos # se actualiza la posición
        mostrar_tablero(posicion) # se redibuja el tablero
        tema = obtener_tema(posicion) # obtiene el tema de la casilla donde cayó
        if preguntar(tema): # llama a preguntar, que devuelve true o false. el if usa directamente ese valor booleano
            if posicion == TAM_TABLERO: # muestra el mensaje de victoria
                print("¡Felicidades! Has llegado a la última casilla y ganaste el juego.")
                break # termina el bucle
            else:
                print("Continúa al siguiente turno.\n") 
        else:
            print("Perdiste este turno, vuelves a la casilla anterior.")
            posicion -= dado # el jugador retrocede a la casilla donde estaba antes del turno
            mostrar_tablero(posicion)
    print("Gracias por jugar. ¡Hasta luego!") # mensaje de despedida
    sys.exit() # cierra el programa. Aquí se usa el módulo sys que se importó al principio

def menu(): # función sin parámetros que gestiona el menú principal
    bienvenida() # se ejecuta una sola vez, antes del bucle, para mostrar las reglas
    while True: # bucle infitio: el menú se vuelve a mostrar hasta que el usuario elija salir
        opcion = input("Teclea 'j' para jugar o 's' para salir: ").strip().lower() 
        # input pide la opción.
        # strip elimina espacios sobrantes
        # lower convierte el texto en minusculas asi valen igual. Se encadenan los dos métodos en una sola línea
        if opcion == 'j': # llama a jugar e inicia la partida. Cuando la partida teermina, el bucle vuelve a mostrar el menú
            jugar()
        elif opcion == 's': # muestra la despedida y sys.exit cierra el programa
            print("¡Gracias por venir! Hasta luego.")
            sys.exit()
        else: # cualquier otra cosa es inválida; muestra un aviso y el bucle vuelve a preguntar
            print("Opción inválida. Por favor ingresa 'j' o 's'.")

if __name__ == "__main__": 
    menu()
    # name es una variable especial que python crea automáticamente
    # main cuando el archivo se ejecuta directamente
    # menu solo se ejecuta si ejecutas este archivo, no si alguien lo importa para usar sus funciones
    
```

```python
import random  # sirve para generar numeros al azar al tirar el dado y elegir las preguntas
import sys # permite interactuar con el sistema, por ejemplo para cerrar el programa con sys.exit()

TAM_TABLERO, CARAS_DADO = 20, 6

# Preguntas divididas en 4 temas, con texto, opciones y índice de respuesta correcta (1-based)
preguntas = {
    "Matemáticas": [ ("¿Cuánto es 8 x 7?", ["54", "56", "64", "49"], 2),
        ("¿Cuál es la raíz cuadrada de 81?", ["7", "8", "9", "10"], 3),
        ("¿Cuánto es el 15% de 200?", ["25", "30", "20", "35"], 2),
        ("Resuelve x: 3x + 5 = 20", ["3", "4", "5", "6"], 3),
        ("¿Cuánto es 12 + 15?", ["25", "27", "28", "30"], 2)  ],
    "Ciencias": [  ("¿A qué temperatura hierve el agua en ºC?", ["90", "95", "100", "105"], 3),
        ("¿Qué planeta es el planeta rojo?", ["Tierra", "Marte", "Júpiter", "Saturno"], 2),
        ("¿Qué gas absorben las plantas?", ["Oxígeno", "Dióxido de carbono", "Nitrógeno", "Hidrógeno"], 2),
        ("¿Qué es H2O?", ["Oxígeno", "Sal", "Agua", "Hidrógeno"], 3),
        ("¿Cuántos huesos tiene un adulto?", ["206", "201", "210", "195"], 1) ],
    "Historia": [  ("¿Quién fue el primer presidente de Estados Unidos?", ["Abraham Lincoln", "Thomas Jefferson", "George Washington", "John Adams"], 3),
        ("¿En qué año terminó la Segunda Guerra Mundial?", ["1944", "1945", "1946", "1947"], 2),
        ("¿Contra quién se construyó principalmente la Gran Muralla China?", ["Mongoles", "Romanos", "Vikingos", "Persas"], 1),
        ("¿Quién descubrió América?", ["Cristóbal Colón", "Leif Erikson", "Vasco da Gama", "Fernando de Magallanes"], 1),
        ("¿Qué imperio construyó el Coliseo?", ["Griego", "Romano", "Egipcio", "Persa"], 2) ],
    "Geografía": [  ("¿Cuál es el río más largo del mundo?", ["Amazonas", "Nilo", "Yangtsé", "Misisipi"], 2),
        ("¿Qué país tiene mayor población?", ["EE. UU.", "China", "India", "Rusia"], 2),
        ("¿En qué cordillera está el Monte Everest?", ["Andes", "Rockies", "Himalaya", "Alpes"], 3),
        ("¿Cuál es la capital de Australia?", ["Sídney", "Melbourne", "Canberra", "Brisbane"], 3),
        ("¿Cuál es el continente más pequeño?", ["Europa", "Antártida", "Oceanía", "América del Sur"], 3) ]}

# Asignar temas a casillas usando patrón cíclico
def obtener_tema_por_casilla(numero_casilla): # la función asigna un tema de preguntas a cada casilla del tablero siguiendo un patrón cíclico
    temas = list(preguntas.keys())  # preguntas es un diccionario cuyas claves son los temas. Preguntas.keys devuelve esas claves y list las convierte en una lista para poder acceder a ellas por posición
    # Casilla 1 no tiene pregunta, empieza en la 2
    indice = (numero_casilla - 2) % len(temas) # numero de casilla desplaza la numeración para que la casilla 2 corresponda a la posición 0 de la lista, la casilla 3 a la posición 1,...
    # % lens(temas): el operador módulo devuelve el resto de la división entre el número de temas. Así, cuando el número supera la cantidad de temas, el índice vuelve a 0 y se crea el ciclo
    return temas[indice] # devuelve el tema que está en la posición calculada

# Mostrar tablero con ficha en posición actual
def mostrar_tablero(posicion): # define como parámetro posición, que es la casilla en la que está el jugador de 1 a 20
    casillas = [] # crea una lista vacía donde se irá guardando el texto de cada casilla
    for i in range(1, 21): # recorre los números del 1 al 20 ya que el 21 no se incluye. Cada valor de i representa una casilla del tablero
        if i == posicion: # si i coincide con la posicion del jugador, se añade X que representa a la ficha
            casillas.append("[X]")
        else:
            casillas.append(f"[{i}]") # si no, se añade el número de la casilla entre corchetes, usando una f-string que inserta el valor de i dentro del texto
    # Mostrar 5 casillas por línea para mejor visualización
    for i in range(0, 20, 5): # aquí range (0, 20, 5) genera los valorees 0, 5, 10 y 15, es decir, saltando de 5 en 5. Cada valor es el índice donde empieza una fila
        print(" ".join(casillas[i:i+5])) 

# Pedir al usuario que responda pregunta sobre tema dado
def hacer_pregunta(tema): # funcion que plantea al jugador una pregunta de opción múltiple sobre un tema y comprueba si acierta
    print(f"\nPregunta del tema: {tema}") # imprime el tema actual. El /n añade un salto de línea antes, para separar visualmente de lo anterior. La f-string inserta el valor de tema en el texto
    pregunta, opciones, correcta = random.choice(preguntas[tema]) #elige una pregunta al azar. Con cada pregunta desempaqueta en tres variables: el texto de la pregunta, la lista de opciones y el número de la opción correcta
    print(pregunta)
    for idx, opcion in enumerate(opciones, 1): # recorre la lista dando a cada opción un número que empieza en 1, 2, 3, 4
        print(f" {idx}. {opcion}")
    while True: # repite indefinidamente hasta que se haga un return
        respuesta = input("Tu respuesta (1-4): ") # pide la respuesta al usuario 
        if respuesta in {"1", "2", "3", "4"}: # se comprueba que esté en el conjunto 1, 2, 3, 4. Se comparan como texto que input devuelve un str
            return int(respuesta) == correcta # si es válida, la convierte a número y la compara con la correcta
        print("Respuesta inválida, escribe un número del 1 al 4.") # si no es válida, se muestra el mensaje de error y el bucle vuelve a pedir la respuesta

# Simular lanzar dado (1-6)
def lanzar_dado(): # función que simula la tirada de un dado de 6 caras
    return random.randint(1, 6) # no recibe parámetros, devuelve un número entero aleatorio entre 1 y 6, ambos incluidos.

# Pantalla de inicio y menú
def pantalla_inicio(): # define función, no recibe parámetros
    print("===== Bienvenido al Juego de Memoria =====") # imprime el título del jueggo.
    while True # repite el menú indefinidamente hasta que se ejecute un return
        print("\nSelecciona una opción:") # el /n deja una línea en blanco antes, para separarlo visualmente
        print(" 1. Jugar")
        print(" 2. Salir")
        opcion = input("Opción (1 o 2): ") # guarda lo que escribe el usuario. Input siempre devuelve texto, por eso compara con 1 y 2 entre comillas
        if opcion == "1": # si elige jugar, la función termina y devuelve true, indicando al programa que debe empezar la partida
            return True
        elif opcion == "2": # si elige salir, muestra un mensaje de despedida y devuelve False, inidcando que el programa tiene que terminar
            print("Gracias por visitar el juego. ¡Hasta luego!")
            return False
        print("Opción inválida, intenta de nuevo.") # esto solo se alcanza si no se ha cumplido ningun return anter, si el usuario escribió otra cosa. Muestra el error y el while true vuelve a mostrar el menu

# Función principal del juego
def juego(): # define la función que controla todo la partida
    jugando = pantalla_inicio()
    if not jugando:
        sys.exit()
# llama a pantalla inicio, que devuelve true (jugar) o false (salir). Si el jugador eligió salir, sys exit termina el programa.
    posicion = 1
    print("\nComienza el juego. Debes llegar a la casilla 20 para ganar.\n")
    mostrar_tablero(posicion)
# La ficha empieza en la casilla 1, se muestra el mensaje de inicio y se dibuja el tablero con la ficha en esa posición
    while posicion < 20: # se repite cada turno mientras el jugador no haya llegado a la casilla 20
        input("\nPresiona Enter para lanzar el dado...")
        dado = lanzar_dado()
        print(f"Has sacado un {dado}.") 
    # el input sirve solo como pausa: el programa espera a que el jugador pulse enter. Despues se obtiene un número al azar del 1 al 6 y se muestra

        posicion += dado # suma la tirada a la posición actual
        if posicion > 20: # si se pasa de 20, se limita a 20, por lo que no hace falta sacar el número exacto para ganar
            posicion = 20
        print(f"Avanzas a la casilla {posicion}.")
        mostrar_tablero(posicion)

        if posicion == 20: # si llego a la casilla 20, se felicita al jugador y break sale del bucle while. Por eso no se hace pregunta en la casilla final
            print("\n¡Felicidades! Has llegado a la última casilla y ganado el juego.")
            break

        # Pregunta al caer en la casilla
        tema = obtener_tema_por_casilla(posicion)
        correcto = hacer_pregunta(tema)
        # se obtiene el tema de esa casilla (función cíclica) y se lanza una pregunta del tema. hacer pregunta devuelve true si acierta y false si falla y se guarda en correcto
        if correcto:
            print("¡Respuesta correcta! Sigue avanzando.")
        else:
            print("Respuesta incorrecta. Pero no te preocupes, sigue intentando.")
        # solo muestra un mensaje distinto según el resultado. Acertar o fallar no cambia la posición la ficha avanza igual

    print("\nGracias por jugar. El juego terminará ahora.") # esta fuera del while, así que se muestra al terminar la pàrtida

if __name__ == "__main__":
    juego()
    # name es una variable especial que python crea automáticamente
    # main cuando el archivo se ejecuta directamente
    # juego solo se ejecuta si ejecutas este archivo, no si alguien lo importa para usar sus funciones
    
```
